# Remittances & Inflation in The Gambia — Exploratory Analysis

Run `scripts/fetch_data.py` first to populate `data/raw/` before running this notebook.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

RAW_DIR = os.path.join('..', 'data', 'raw')

remit = pd.read_csv(os.path.join(RAW_DIR, 'remittances_pct_gdp.csv'))
cpi = pd.read_csv(os.path.join(RAW_DIR, 'cpi_inflation_pct.csv'))

df = pd.merge(remit, cpi, on='year', how='inner').dropna().sort_values('year')
df.head()
```

```python
fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.set_xlabel('Year')
ax1.set_ylabel('Remittances (% of GDP)', color='tab:blue')
ax1.plot(df['year'], df['BX.TRF.PWKR.DT.GD.ZS'], color='tab:blue', marker='o')
ax1.tick_params(axis='y', labelcolor='tab:blue')

ax2 = ax1.twinx()
ax2.set_ylabel('CPI Inflation (annual %)', color='tab:red')
ax2.plot(df['year'], df['FP.CPI.TOTL.ZG'], color='tab:red', marker='s')
ax2.tick_params(axis='y', labelcolor='tab:red')

plt.title('The Gambia: Remittances vs. CPI Inflation')
fig.tight_layout()
plt.show()
```

```python
corr = df['BX.TRF.PWKR.DT.GD.ZS'].corr(df['FP.CPI.TOTL.ZG'])
print(f'Correlation: {corr:.3f}')
```

```python
import statsmodels.api as sm

X = sm.add_constant(df['BX.TRF.PWKR.DT.GD.ZS'])
y = df['FP.CPI.TOTL.ZG']
model = sm.OLS(y, X).fit()
print(model.summary())
```
